```python
#Functions — packaging code you can reuse#
```

```python
def check_supply(demand, supply):
    if demand > supply:
        print("Shortage")
    elif demand == supply:
        print("Balanced")
    else:
        print("Sufficient")
```

```python
#demand and supply in the parentheses are called parameters — placeholders for values you'll provide later#
```

```python
check_supply(150, 130)
```

```python
check_supply(100, 130)
```

```python
check_supply(130, 130)
```

```python
#Right now, check_supply only prints — it doesn't give you anything you can store or reuse. Let's fix that with return:#
```

```python
def get_gap(demand, supply):
    gap = demand - supply
    return gap
```

```python
result = get_gap(150, 130)
result
```

```python
#default values for parameters#
```

```python
#For example, say most of the time you're checking against a fixed benchmark supply, and typing it every single time is repetitive.#
```

```python
def check_supply(demand, supply=130):
    if demand > supply:
        print("Shortage")
    elif demand == supply:
        print("Balanced")
    else:
        print("Sufficient")
```

```python
check_supply(150)
```

```python
check_supply(150, supply=200)
```
