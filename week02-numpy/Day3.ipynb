```python
# 2D arrays — rows and columns#
# real energy data usually looks like a table — rows and columns (e.g., each row a day, each column an hour).#
# NumPy handles this too, using what's called a 2D array#
```

```python
import numpy as np

demand_table = np.array([
    [100, 120, 140, 110],
    [90, 130, 150, 100],
    [95, 125, 145, 105]
])
demand_table
```

```python
demand_table.shape
```

```python
# Accessing a specific value#
```

```python
demand_table[0, 2]
```

```python
# Grabbing a whole row or column#
```

```python
demand_table[0]
```

```python
demand_table[:, 1]
```
