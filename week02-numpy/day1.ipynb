```python
#NumPy is a library (a toolbox of extra code someone else wrote) built specifically for fast, easy math on lots of numbers at once. 
It gives you a new type called an array, which looks similar to a list but behaves very differently when you do math on it#
```

```python
import numpy as np
```

```python
daily_demand = np.array([120, 135, 130, 140, 150, 145])
daily_demand
```

```python
daily_demand / 1000
```

```python
#Exactly right — and notice what just happened: one line, no loop, and NumPy divided every single value by 1000 all at once#
#This is called vectorization — doing math on a whole array in a single operation, instead of looping through it one item at a time#
```

```python
daily_demand + 10
```

```python
daily_demand.mean()
```

```python
daily_demand.max()
```

```python
daily_demand.min()
```

```python
daily_demand.sum()
```

```python
# Slicing means to grab a range of items using a colon: #

#This means "give me items from position 0 up to (but not including) position 3."#
```

```python
daily_demand[0:3]
```

```python
daily_demand[3:]
```
