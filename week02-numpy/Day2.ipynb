```python
#Filtering — picking out values that meet a condition#
```

```python
import numpy as np

daily_demand = np.array([120, 135, 130, 140, 150, 145])

daily_demand > 135
```

```python
daily_demand[daily_demand > 135]
```

```python
#Combining conditions#

#With NumPy arrays, you use & (and) and | (or) instead — and each condition needs its own parentheses#
```

```python
daily_demand[(daily_demand > 125) & (daily_demand < 145)]
```

```python
daily_demand[(daily_demand > 125) | (daily_demand < 145)]
```
