```python
#if statements/conditionals#
```

```python
demand = 150
supply = 130

if demand > supply:
    print("Shortage")
```

```python
demand = 100
supply = 130

if demand > supply:
    print("Shortage")
else:
    print("Sufficient supply")
```

```python
demand = 130
supply = 130

if demand > supply:
    print("Shortage")
elif demand == supply:
    print("Balanced")
else:
    print("Sufficient supply")
```

```python
#combining conditionals with 'for' loop#
```

```python
daily_demand = [120, 135, 130, 140, 150, 145]
daily_supply = 135

for day in daily_demand:
    if day > daily_supply:
        print(day, "- Shortage")
    elif day == daily_supply:
        print(day, "- Balanced")
    else:
        print(day, "- Sufficient")
```

```python
#combining more than one condition at once#
```

```python
demand = 150
supply = 130
is_weekday = True

if demand > supply and is_weekday:
    print("Critical shortage")
else:
    print("Not critical")
```

```python
demand = 150
supply = 130
is_weekday = False

if demand > supply and is_weekday:
    print("Critical shortage")
else:
    print("Not critical")
```
